## 2. Data Cleaning

```python
def parse_mixed_date(value):
    """Try both date formats present in the raw file and return a proper Timestamp."""
    for fmt in ('%m-%d-%Y', '%m/%d/%Y'):
        try:
            return pd.to_datetime(value, format=fmt)
        except ValueError:
            continue
    return pd.NaT

df_clean = df.copy()
df_clean['date'] = df_clean['date'].apply(parse_mixed_date)

print(f"Dates successfully parsed : {df_clean['date'].notna().sum()} / {len(df_clean)}")
print(f"Date range                : {df_clean['date'].min().date()}  to  {df_clean['date'].max().date()}")
```

```python
print("Before cleaning:")
print(df_clean['transaction_type'].value_counts(dropna=False))

fill_values = {
    'transaction_type': 'Cash'
}

df_clean = df_clean.fillna(fill_values)

print("\nAfter cleaning:")
print(df_clean['transaction_type'].value_counts())
```

```python
print(df_clean.isnull().sum())
```

```python
print(f"Duplicate rows: {df_clean.duplicated().sum()}")
print(f"\nUpdated dtypes:")
print(df_clean.dtypes)
```

## Data Processing


```python
# Extract date-related features
df_clean = df_clean.assign(
    month=df_clean['date'].dt.strftime('%B'),
    month_num=df_clean['date'].dt.month,
    weekday=df_clean['date'].dt.strftime('%A'),
    quarter=df_clean['date'].dt.quarter
)

# Check whether transaction amount matches price multiplied by quantity
df_clean['calculated_amount'] = df_clean.eval('item_price * quantity')

mismatches = df_clean.eval(
    'calculated_amount != transaction_amount'
).sum()

print("Rows where transaction_amount != price*quantity:", mismatches)

# Remove the temporary column
df_clean.drop('calculated_amount', axis=1, inplace=True)

# Display selected columns
print(df_clean.loc[:, ['date', 'month', 'weekday', 'quarter']].head())
```

```python
# Calculate total revenue and quantity for each product
revenue_by_item = (
    df_clean.pivot_table(
        index='item_name',
        values='transaction_amount',
        aggfunc='sum'
    )
    .squeeze()
    .sort_values(ascending=False)
)

quantity_by_item = (
    df_clean.pivot_table(
        index='item_name',
        values='quantity',
        aggfunc='sum'
    )
    .squeeze()
    .sort_values(ascending=False)
)

print("Revenue by product (₹):")
print(revenue_by_item)
```

```python
file_name = 'balaji_fast_food_cleaned.csv'

df_clean.to_csv(file_name, index=False)

print("Cleaned dataset saved as:", file_name)
print("Final shape:", df_clean.shape)
```

## 3. Data Visualization

### 3.1 Which products generate the highest revenue

```python
# Create revenue chart
ax = revenue_by_item.plot(
    kind='barh',
    figsize=(10, 6),
    title='Total Revenue by Product'
)

ax.set_xlabel('Revenue (₹)')
ax.set_ylabel('Product')

for i, value in enumerate(revenue_by_item):
    ax.text(value, i, f'₹{value:,.0f}', va='center')

plt.tight_layout()
plt.show()
```

### 3.2 Total Quantity Sold by Product

```python
# Plot total quantity sold for each product
ax = quantity_by_item.plot(
    kind='barh',
    figsize=(10, 6),
    title='Total Quantity Sold by Product'
)

ax.set_xlabel('Units Sold')
ax.set_ylabel('Product')

plt.tight_layout()
plt.show()
```

### 3.3 Fastfood Vs Beverages in terms of Revenue

```python
# Calculate revenue for each category
revenue_by_category = df_clean.groupby('item_type')['transaction_amount'].sum()

ax = revenue_by_category.plot.pie(
    figsize=(7, 7),
    autopct='%1.1f%%',
    startangle=90
)

ax.set_title("Revenue Contribution by Category")
ax.set_ylabel("")

plt.tight_layout()
plt.show()
```

### 3.4 Sales Revenue Across Different Times In The Day

```python
time_order = ['Morning', 'Afternoon', 'Evening', 'Night', 'Midnight']

revenue_by_time = (
    df_clean.groupby('time_of_sale')['transaction_amount']
    .sum()
    .reindex(time_order)
)

# Plot revenue by time of sale
ax = revenue_by_time.plot(
    kind='bar',
    figsize=(10, 6),
    title='Sales Revenue Across Different Times'
)

ax.set_xlabel('Time of Sale')
ax.set_ylabel('Revenue (₹)')

plt.tight_layout()
plt.show()
```

### 3.5  Revenue trend across the year

```python
# Create Month-Year column
df_clean['month_year'] = df_clean['date'].dt.strftime('%b %Y')

# Arrange months in the correct order
month_order = pd.date_range(
    start=df_clean['date'].min(),
    end=df_clean['date'].max(),
    freq='MS'
).strftime('%b %Y')

# Calculate monthly revenue
revenue_by_month = (
    df_clean.groupby('month_year')['transaction_amount']
    .sum()
    .reindex(month_order)
)

# Plot the trend
plt.figure(figsize=(12, 6))

plt.plot(
    revenue_by_month.index,
    revenue_by_month.values,
    marker='o',
    linewidth=2
)

plt.title("Monthly Revenue Trend")
plt.xlabel("Month")
plt.ylabel("Revenue (₹)")
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

### 3.6 Payment Method Most of The Customers prefer

```python
# Count each payment method
payment_counts = df_clean['transaction_type'].value_counts()

# Create pie chart
payment_counts.plot(
    kind='pie',
    figsize=(7, 7),
    autopct='%1.1f%%',
    startangle=90
)

plt.title('Payment Method Distribution')
plt.ylabel('')

plt.tight_layout()
plt.show()
```

### 3.7 Revenue By Product And Time Of Sale

```python
# Revenue by product and time
pivot = df_clean.pivot_table(
    index='item_name',
    columns='time_of_sale',
    values='transaction_amount',
    aggfunc='sum'
).reindex(columns=time_order)

plt.figure(figsize=(9, 6))

sns.heatmap(
    pivot,
    annot=True,
    fmt='.0f',
    cmap='YlOrRd'
)

plt.title("Revenue by Product and Time of Sale")
plt.xlabel("Time of Sale")
plt.ylabel("Product")

plt.tight_layout()
plt.show()
```

### 3.8 Correlation Analysis

```python
# Calculate correlation between numerical columns
corr_matrix = df_clean[
    ['item_price', 'quantity', 'transaction_amount']
].corr()

# Create heatmap
plt.figure(figsize=(6, 5))

sns.heatmap(
    corr_matrix,
    annot=True,
    fmt='.2f',
    cmap='coolwarm',
    center=0
)

plt.title("Correlation Between Sales Variables")
plt.tight_layout()
plt.show()
```

### 3.9 Which staff members handle the most revenue

```python
# Calculate total revenue by staff
revenue_by_staff = (
    df_clean.groupby('received_by')['transaction_amount']
    .sum()
)

# Plot revenue by staff
revenue_by_staff.plot(
    kind='bar',
    figsize=(5,5),
    title='Revenue by Staff'
)

plt.xlabel('Staff')
plt.ylabel('Revenue (₹)')

plt.tight_layout()
plt.show()
```

### 3.10 Revenue by Day of  Week

```python
# Create weekday column
df_clean['weekday'] = df_clean['date'].dt.day_name()

# Set weekday order
weekday_order = [
    'Monday', 'Tuesday', 'Wednesday',
    'Thursday', 'Friday', 'Saturday', 'Sunday'
]

# Calculate weekday-wise revenue
revenue_by_weekday = (
    df_clean.groupby('weekday')['transaction_amount']
    .sum()
    .reindex(weekday_order)
)

# Plot revenue by weekday
revenue_by_weekday.plot(
    kind='bar',
    figsize=(10, 6),
    title='Revenue by Day of Week'
)

plt.xlabel('Day of Week')
plt.ylabel('Revenue (₹)')
plt.xticks(rotation=30)

plt.tight_layout()
plt.show()
```

### 3.11 Number of Orders By Product

```python
# Count orders for each product
order_count = df_clean['item_name'].value_counts()

# Plot the order count
order_count.plot(
    kind='barh',
    figsize=(10, 6),
    title='Number of Orders by Product'
)

plt.xlabel('Number of Orders')
plt.ylabel('Product')

plt.tight_layout()
plt.show()
```

- **Sandwich** generated the highest revenue of **₹65,820**.
- **Cold Coffee** had the highest quantity sold with **1,361 units**.
- **Vadapav** generated the lowest revenue at **₹20,120**.
- A product with high sales quantity does not always generate the highest revenue.

### 4.4. Category-wise Revenue

- **Fastfood** generated **₹1,88,840**, contributing **68.6%** of total revenue.
- **Beverages** generated **₹86,390**, contributing **31.4%**.
- Fastfood items are therefore the major source of revenue.

### 4.5. Payment Analysis

- **Cash** accounted for **47.6%** of transactions.
- **Online** payments accounted for **41.7%**.
- Some payment records were missing, so complete payment information would improve future analysis.

### 4.6. Time and Staff Performance

- **Night** generated the highest revenue at **₹62,075**.
- **Midnight** generated **₹50,725**.

Overall, **Fastfood items are the main revenue contributors**, while **Cold Coffee is the best-selling product by quantity**.
